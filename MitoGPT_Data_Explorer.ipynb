```python
# === STEP 1: Mount Google Drive ===
from google.colab import drive
drive.mount('/content/drive')

# === STEP 2: Define your main folder path ===
# Update this path if MitoGPT is inside another directory
import os
root_path = "/content/drive/MyDrive/MitoGPT"

# === STEP 3: Function to print tree structure ===
def print_tree(startpath, prefix=""):
    """
    Recursively prints the directory tree of a folder.
    Shows folder hierarchy and file sizes.
    """
    items = sorted(os.listdir(startpath))
    for i, name in enumerate(items):
        path = os.path.join(startpath, name)
        connector = "└── " if i == len(items)-1 else "├── "
        if os.path.isdir(path):
            print(prefix + connector + f"[DIR] {name}/")
            extension = "    " if i == len(items)-1 else "│   "
            print_tree(path, prefix + extension)
        else:
            size = os.path.getsize(path) / 1024  # KB
            print(prefix + connector + f"{name} ({size:.1f} KB)")

# === STEP 4: Run it ===
print(f"\n📂 Folder structure for: {root_path}\n")
print_tree(root_path)
```
